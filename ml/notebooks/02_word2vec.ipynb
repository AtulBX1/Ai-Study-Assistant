# Unit II: Word2Vec experiments

This lightweight notebook compares CBOW and Skip-Gram on a tiny in-memory corpus. It does not download models or save large outputs. Tiny examples illustrate parameters; they are not suitable for production embeddings.

```python
from gensim.models import Word2Vec

sentences = [
    "king queen royal crown kingdom".split(),
    "man woman king queen person".split(),
    "queen woman royal crown ruler".split(),
    "biology cell gene protein organism".split(),
    "cell gene biology organism protein".split(),
]
```

## Hyperparameters

- `vector_size` is the embedding dimension. A larger vector has more capacity but needs more examples.
- `window` is the maximum number of context words to either side. Smaller windows emphasize nearby syntactic context; larger windows capture broader topical context.
- `min_count` discards words occurring fewer times than the threshold. Keep it at 1 for this small corpus.
- `epochs` is the number of passes over the corpus. More passes mean more updates, but can overfit tiny data.
- `negative` is the number of sampled negative context predictions per positive example.
- `seed` controls random initialization, while `workers=1` makes this small comparison more repeatable.

```python
settings = dict(vector_size=30, window=2, min_count=1, negative=5, seed=42, workers=1, epochs=30)
cbow = Word2Vec(sentences=sentences, sg=0, **settings)
skipgram = Word2Vec(sentences=sentences, sg=1, **settings)

print("CBOW vocabulary:", len(cbow.wv))
print("Skip-Gram vocabulary:", len(skipgram.wv))
print("CBOW neighbours of king:", cbow.wv.most_similar("king", topn=3))
print("Skip-Gram neighbours of king:", skipgram.wv.most_similar("king", topn=3))
```

## Small parameter comparisons

Try increasing `window`, then compare vectors. Changing `min_count` to 2 removes singleton words if there are any. A larger `vector_size` is not automatically better: on a small corpus it increases parameters without adding evidence. `negative` changes how many contrastive samples are trained for each observed word-context pair.

```python
wide_window = Word2Vec(
    sentences=sentences,
    sg=0,
    vector_size=30,
    window=4,
    min_count=1,
    negative=2,
    seed=42,
    workers=1,
    epochs=30,
)
print("Window 2:", cbow.wv.most_similar("king", topn=3))
print("Window 4:", wide_window.wv.most_similar("king", topn=3))
```
